1. What is the difference between a function and a method in Python?
In Python, functions and methods are both blocks of reusable code, but they have a key difference in how they are associated with objects:

1. Function
A function is an independent block of code that performs a specific task.

It is not tied to any object.

Defined using the def keyword outside of a class.

Can be called directly using its name.

ex:
def greet(name):
    return f"Hello, {name}!"

greet("Alice")  # Function call

2. Method
A method is a function that is associated with an object, specifically an instance of a class.

Defined inside a class and automatically takes self as the first argument, representing the instance.

Called on an object (instance of the class).

ex:
class Greeter:
    def greet(self, name):
        return f"Hello, {name}!"

g = Greeter()
g.greet("Alice")  # Method call on object `g`

3. What are the different ways to define and call a function in Python?


In Python, you can define and call functions in several ways depending on how flexible or complex the behavior needs to be. Here’s a breakdown:

> 1. Standard Function Definition
Defined using def, called using its name.

def greet(name):
    return f"Hello, {name}!"

print(greet("Alice"))  # Standard function call

> 2. Function with Default Arguments

def greet(name="Guest"):
    return f"Hello, {name}!"

print(greet())          # Uses default
print(greet("Bob"))     # Overrides default

> 3. Function with Variable-length Arguments

def add_numbers(*args):
    return sum(args)

print(add_numbers(1, 2, 3, 4))

def print_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_info(name="Alice", age=30)

> 4. Lambda (Anonymous) Functions

square = lambda x: x ** 2
print(square(5))

> 5. Calling Functions with Keyword Arguments

def greet(name, message):
    return f"{message}, {name}!"

print(greet(name="Alice", message="Good morning"))

> 6. Nested Functions

def outer():
    def inner():
        return "Inner function"
    return inner()

print(outer())

> 7. Function as Arguments / First-Class Functions

def shout(text):
    return text.upper()

def whisper(text):
    return text.lower()

def speak(func, message):
    return func(message)

print(speak(shout, "Hello"))
print(speak(whisper, "Hello"))

> 8. Calling Methods (Functions inside Classes)

class Greeter:
    def greet(self, name):
        return f"Hello, {name}!"

g = Greeter()
print(g.greet("Alice"))  # Method call

Iterators vs Iterables in Python

Must implement the __iter__() method.

ex: my_list = [1, 2, 3]      # This is an iterable

    print(item)

An object that produces values one at a time using __next__().

Maintains internal state and knows where it is during iteration.
 ex: my_list = [1, 2, 3]
it = iter(my_list)         # Create iterator from iterable

print(next(it))  # 1

6. Explain the concept of generators in Python and how they are defined? with example

> Generators in Python
Generators are a special type of iterator that are defined using functions but allow you to generate values on the fly using the yield keyword instead of return.

They are memory-efficient and ideal for working with large data or streams.

Key Features of Generators:
Use yield instead of return
Automatically create an iterator
Maintain internal state between calls
Pause and resume execution

How to Define a Generator
def count_up_to(n):
    i = 1
    while i <= n:
        yield i
        i += 1

> Using the Generator
counter = count_up_to(3)

print(next(counter))  # 1
print(next(counter))  # 2
print(next(counter))  # 3
# print(next(counter))  # Raises StopIteration

> Using in a Loop
for num in count_up_to(3):
    print(num)
# Output: 1, 2, 3

7. What are the advantages of using generators over regular functions? with example

Advantages of Using Generators Over Regular Functions in Python
Generators provide several practical benefits, especially when working with large datasets or streams. Here's a breakdown:

> 1. Memory Efficiency
Generators don't store the entire output in memory.
They yield items one at a time, which is ideal for processing large sequences
def generate_numbers(n):
    for i in range(n):
        yield i

gen = generate_numbers(1000000)  # Uses very little memory

> 2. Lazy Evaluation
Values are computed only when needed, saving processing time.

def lazy_squares(n):
    for i in range(n):
        yield i * i

for square in lazy_squares(5):
    print(square)  # Each value is computed as the loop runs

> 3. Infinite Sequences
Generators can represent infinite or long-running sequences without running out of memory.
 def infinite_counter():
    i = 0
    while True:
        yield i
        i += 1
> 4. Pipeline Support (Chaining Generators)
 Generators can be chained together to create processing pipelines (like Unix pipes).

def numbers():
    for i in range(10):
        yield i

def even_numbers(nums):
    for num in nums:
        if num % 2 == 0:
            yield num

for n in even_numbers(numbers()):
    print(n)

10. What is the difference between map(), reduce(), and filter() functions in Python? with example

> Difference Between map(), filter(), and reduce() in Python
These are all functional programming tools in Python, used to process collections (like lists or tuples). Here's how they differ:

> map()
Purpose: Applies a function to each item in an iterable and returns an iterator of results.

Example:
nums = [1, 2, 3, 4]
squares = list(map(lambda x: x ** 2, nums))
print(squares)  # Output: [1, 4, 9, 16]

> filter()
Purpose: Filters items from an iterable based on a condition (Boolean function).
Only elements for which the function returns True are kept.

nums = [1, 2, 3, 4, 5]
evens = list(filter(lambda x: x % 2 == 0, nums))
print(evens)  # Output: [2, 4]

> reduce() (from functools module)
Purpose: Applies a function cumulatively to the items in an iterable, reducing them to a single value.

from functools import reduce

nums = [1, 2, 3, 4]
product = reduce(lambda x, y: x * y, nums)
print(product)  # Output: 24

```python
# 4. Write a Python function that checks if a given number is prime or not from 1 to 200.
def is_prime(n):
    if n <= 1:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False
    for i in range(3, int(n ** 0.5) + 1, 2):
        if n % i == 0:
            return False
    return True

for num in range(1, 201):
    if is_prime(num):
        print(f"{num} is a prime number")
```

```python
# 5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.
class FibonacciIterator:
    def __init__(self, max_terms):
        self.max_terms = max_terms
        self.count = 0
        self.a = 0
        self.b = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.max_terms:
            raise StopIteration
        if self.count == 0:
            self.count += 1
            return self.a
        elif self.count == 1:
            self.count += 1
            return self.b
        else:
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return self.b

fib = FibonacciIterator(10)

for num in fib:
    print(num, end=' ')
```

```python
# 6. Write a generator function in Python that yields the powers of 2 up to a given exponent.
def powers_of_two(max_exponent):
    for exp in range(max_exponent + 1):
        yield 2 ** exp

for power in powers_of_two(5):
    print(power)
```

```python
# 9. Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.
# List of temperatures in Celsius
celsius = [0, 20, 37, 100]

# Function to convert to Fahrenheit
def to_fahrenheit(c):
    return (c * 9/5) + 32

# Use map to apply conversion
fahrenheit = list(map(to_fahrenheit, celsius))

# Output result
print("Temperatures in Fahrenheit:", fahrenheit)
```

```python
# 10. Create a Python program that uses `filter()` to remove all the vowels from a given string.
# Function to check if a character is not a vowel
def is_not_vowel(char):
    return char.lower() not in 'aeiou'

# Input string
input_string = "Hello World!"

# Use filter to remove vowels
result = ''.join(filter(is_not_vowel, input_string))

# Output result
print("String after removing vowels:", result)
```
